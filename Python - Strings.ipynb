### Definition: 
A data type used to represent text in Python and it contain letters, numbers, symbols, spaces and other characters. 

```python
# Example: 
name = "Yubaraj"
message = 'Hello, World!'
```

Note* To reprsent text in Python, computer cannot directly understand letter such as A, B, or, Hello. Internally, text is represented as a sequence of characters and each characters is stored using a numerical representation based on unicode.

### String Length:

```python
# Example: 
name = "Yubaraj"
print(len(name))
```

### Character Access and Indexing:


 Indexing: 

Each characters in the string has a position called a index, starting from 0, not 1.as_integer_ratio.

```python
# Example: 
text = "Python" 
# here, Character: P y t h o n
#       Index:     0 1 2 3 4 5
```

Accessing: Use [].

```python
# Example: 
text = "Yubaraj"
print(text[0])
print(text[3])
print (text[5]) 
```

### Positive and Neagtive Indexing:

```python
# Example: 
text = 'Python'
# Here, text :  P  y  t  h  o  n 
#       Index:  0  1  2  3  4  5
#    Neagtive: -6 -5 -4 -3 -2 -1 
print (text[2])
print (text[4])
print (text[-1])
print(text[-5])
```

### String Slicing: 

It means extracting a part of string instead of accessing only one character; using index value. 

```python
# Example: 
text = "Python"
print(text[0:3])
```

1. Omitting the Start Index: 

```python
# Example: 
text = "Python"
print(text[:4])
```

2. Omitting the Stop Index:

```python
# Example: 
text = "Hello"
print(text[2:])
```

3. Neagtive Slicing:

```python
# Example: 
text = "Kathmandu"
print(text[-4:-1])
```

4. Reversing String:

```python
# Example: 
text = "Kathmandu"
print(text[::-1])
```

### String Operation: 


1. String Concatenation: Joining two or more string using +

```python
# Example: 
first_name = "Mann"
last_name = "Shrestha"
full_name = first_name + " " + last_name 
print(full_name)
```

2. String Repetition: 

```python
# Example: 
text = "Hi"
print (text *3)
```

3. String Lenght: Using len() Function:

```python
# Example: 
text = "Kathmandu"
print(len(text))
```

4. Checking whether Text exit or not: using in operator:

```python
# Example: 
text = "Python Programming"
print ("Python" in text)
print ("Java" in text)
```

### String Method: 


1. Upper Case: .upper() function

```python
# Example: 
text = "Kathmandu"
print (text.upper())
```

2. Lower Case: .lower() Function

```python
# Example: 
text = "Kathmandu"
print(text.lower())
```

3. Capitilized Case: .capitilized() Function

```python
# Example: 
text = "mann shrestha"
print(text.capitalize())
```

4. Title Case: .title() Function

```python
# Example: 
text = "jane foster"
print(text.title())
```

5. Strip Case: .strip() Function; It remove the space from beginning and end of string

```python
# Example:
text = " Hello Yubaraj "
print (text.strip())
```

6. Replace Case: .replace() Function; It replace one piece of text with another.

```python
# Example: 
text = "I love you Jannu"
print (text.replace("Jannu", "Baby"))
```

7. Find Method: .find() Function

```python
# Example: 
text = "I live in Kathmandu and sometime i live in Vienna."
print(text.find("Vienna"))
```

8. Count Method: .count() Function; How many times something appears

```python
# Example: 
text = 'Banana'
print(text.count("a"))
```

9. Startwith Method: .startswith() Function; Find something start with in result True or False

```python
# Example: 
text = "Python Programming"
print(text.startswith("Programming"))
```

10. Endswith Method: .endswith() Function; Find something end with in result True or Flase

```python
# Example: 
text = "Python Programming"
print(text.endswith("Programming"))
```

11. Spilt Method: .split() Function; Breaks a string into a smaller strings

```python
# Example: 
text = "Python is Programming language"
word = text.split()
print(word)
```

12. Join Method: .join() Function; Joins multiple string together

```python
# Example: 
text = ['Python', 'is', 'Programming', 'language']
word = " ".join(text)
print (word)
```

### The Immutable Nature of String: 

In Python, String are immutable, which means that once it is created, its individual characters cannot be changed directly.

```python
# Example: 
text = "I love Vienna"
new_text = text.replace("Vienna", "London")
print (text)
print (new_text)

# note* the original string remain unchanged, instead it print new string. 
```
