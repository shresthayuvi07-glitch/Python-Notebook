### Definition:
 It is a name used to store data in memory and acts as a container that holds a value, which can be changed during the execution of a program.

```python
# Example:
name = "Yubaraj"
sur_name = "Shrestha"
age = 24
height = 5.11
is_human = True
# Here:
# name and sur_name         - store a string
# age                       - store an integer
# height                    - store a floating point number
# is_human                  - store a Boolean Value.
```

### Rules for naming Variable:

1. Must start with a letter or underscore (_)
2. Cannot start with a number
3. Cannot use Python keywords (Eg, if, for, while)
4. Variables name are case sensitive (Age and age are different)

### Changing Variables Value:

1. Changing a Number:

```python
# Example: 
age = 20 
print(age)

age = 25
print(age)
```

2. Changing a String: 

```python
# Example: 
name = "Yubaraj"
print(name)

name = "Mann"
print(name)
```

3. Changing a Value using Calculation:

```python
# Example: 
score = 50
score = score + 10
print(score)
```

4. Using different Arthematic Operators:


```python
# Example:
x = 10

x += 5      # x = 15
x -= 3      # x = 12
x *= 2      # x = 24
x /= 4      # x = 6.0
print(x)
```
