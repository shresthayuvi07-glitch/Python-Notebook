## Python_Print_Statement Notebook:

### The purpose of displaying output:


1. To show Result
2. To display information
3. To show Variables Value
4. To give Feedback to User
5. To help find Errors
6. To display Multiple Value
7. To make programs Interactive 

### Printing Texts, Numbers, Variables and Calculated results:

1. Texts: 

```python
# Example: 
print("Yubaraj Shrestha")
print("Hello Yubaraj")
```

2. Numbers:

```python
# Example: 
print(100)
print(150)
print(25.5)
```

3. Variables:


```python
# Example: 
name = "Yubaraj"
age = 24

print(name)
print(age)
```

4. Calculated Result:


```python
# Example: 
print(10+5)
print(30-15)
print(3*15)
print(30/2)
```

### Displaying Multiple Values: 


```python
# Example:
name = "Yubaraj"
age = 24 
height = 5.11

print("Name:", name, "Age:", age, "Height", height)
```

### Displaying Using F-string:

```python
# Example: 
name = "Yubaraj"
age = 24
place = "Kathmandu"

print(f"My name is {name} and I am {age} years old. I live in {place}.")
```

### Output Separator (sep=" "): To separate the output.

```python
# Example:
print("Python", "SQL", "Power BI")    # Whithout output Separator 

print("Python", "SQL", "Power BI", sep=", ") # With output separator
print("Python", "SQL", "Power BI", sep="/ ")
```

### Output Ending Behavior (end=" ") or, (\n):


```python
# Example: 
print("Hello")    # Without output ending behavior
print("Yubaraj")

print("Hello", end=" ")   # With output ending behavior
print("Yubaraj")

print("Hello", end="")
print("Yubaraj")

print ("Hello", end="-")
print ("Yubaraj")
```

### Output formatting changes based on different setting:


1. Using Output Separator
2. using Output End Behavior 
3. Formatting Decimal numbers 
4. Adding commas to large numbers
5. Controlling width and alignment


```python
# Example: 
name = "Yubaraj"
price = 1200.5078

print("Name", name, sep=":", end="|")
print("Price", price, sep=":")
```

### Common Output-related Mistake:

1. 1. Forgetting quotation marks for text


```python
# Example: 
#print(Hello)   # wrong

print("Hello")   # Correct
```

2. Forgetting Parentheses with print()

```python
# Exmaple:
#print "Hello"  # Wrong
print ("Hello World")
```

3. Using an Undifine variables

```python
# Exmaple:
age = 24
#print("name")      # Wrong
print(age)
```

4. Confusing text with variables

```python
# Exmaple:
name = "Yubaraj"
print ("name")    # Confusion of text with variables
# or, 
print (name)
```

5. Incorrect Capitalization


```python
# Example:
Name = "Mann"
#print(name) 

# or, 
Name = "Mann"
print(Name)
# Name and name are diffirent variables
```

6. Incorrect use of Commas and +


```python
# Example:
name = "Yubaraj"
age = 24
print ("Name:" + name, "Age:", age)
# note* Right way of using commas and +
```

7. Forgetting space when joining string


```python
# Example:
first = "Hello"
second = "Mann"
print (first +  second)   # Mistake
print (first, "", second)
```

8. Incorrect Indentation


```python
# Example:
#if True: 
#print("Hello")     #Wrong

if True:
    print("Hello")
```

9. Using wrong Separator

```python
# Example:
print("A", "B", "C")   # without

print ("A", "B", "C", sep=",")  # with separator
```

10. Using wrong print ends


```python
# Example: 
print ("Hello")
print ("Mann")      # Without

print ("Hello", end=" ")
print ("Manm")      # with print end
```
