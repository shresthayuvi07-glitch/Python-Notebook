## Python_Data_Types Notebook:

### Definition: 
Data Types tells Python what kind of value a Variables contains and different types of data can be used for different purpose, such as storing numbers, text, or True/False values. 



### They Matters for:
1. They determine what operation you can perform
2. They help python handle data correctly
3. They prevent incorrect operations 
4. They affect memory and program behavior

### Types of Data in Python:

1. String: 


```python
# Example: 
name = "Mann"
city = "Kathmandu" 
print (name)
print (city)
```

2. Integer: 


```python
# Example: 
age = 20 
quantity = 50 
print (age)
print (quantity)
```

3. Float: 

```python
# Example: 
price = 99.50
height = 5.10
print (price)
print (height)
```

4. Boolean (bool):


```python
# Example: 
is_logged_in = True
is_admin = False
print(is_logged_in)
print(is_admin)
```

5. List:


```python
# Example: 
fruits = ["Apple", "Orange", "Cherry"]
print (fruits)
```

6. Tuple: 


```python
# Example: 
coordinates = (10, 20)
print (coordinates)
```

7. Dictionary:


```python
# Example: 
student = {
    "name": "Mann", 
    "age": 25,
    "course": "Excel"
}
print (student["name"])
```

8. Set: 


```python
# Example: 
numbers = {10, 20, 30, 40, 10}
print(numbers)
```

### Identifies the types of Values: 

```python
# Example: 
name = "Mann"
age = 24
price = 99.90
student = True
print (type(name))
print (type(age))
print (type(price))
print (type(student))
```

### Comparing Data Types:


```python
# Example: 
age = 20
marks = 80 
print (type(age) == type (marks))       # Integer and Integer

age = 24
marks = 80.5
print (type(age)==type(marks))      # Float and Integer
```

### Same operation but different behavior based on data type:

1. (+) with Integer: it performs addition.

```python
# Example: 
a = 10
b = 5 
print (a+b)
```

2. (+) with String: it performs concatenation (joining text).

```python
# Example: 
a = "30"
b = "20"
print(a+b)
```

3. (+) with List: it also join lists.

```python
# Example: 
list1 = [1, 2, 3]
list2 = [7, 8, 9]
print (list1 + list2)
```

4. (*) with Integer: it performs multiplication.

```python
# Example: 
a = 10
b = 5 
print(a*b)
```

5. (*) with String: it performs repitition of string.

```python
# Example:
a = "Hello" 
print(a*3)
```

6. (*) with Lists: it performs repitition of lists.

```python
# Example: 
list = [1, 2, 3]
print(list * 2)
```

7. (/) with Integer: it performs division, normally produce float.

```python
# Example: 
a = 10
b = 2 
print(a/b)
```

8. (//) with Integer: it performs floor division.

```python
# Example: 
print (10//3)
```

9. Comparision Operater:

```python
# For Integer: 
# Example: 
print (10>5)

# For String: 
# Example: 
print ("Apple" == "Apple")
```
