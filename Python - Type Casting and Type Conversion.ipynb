## Type Casting and Type Conversion Notebook: 


### Type Conversion 




### Definition: 
Type conversion is the process of changing a value from one data type to another.

### It's needed: 

1. To perform mathematical calculation:

```python
# Example: 
num1 = input("Enter your first number:")
num2 = input("Enter your second number:")
print(num1 + num2)
# note* This happens because Python is joining two strings

# Convert into integers: 
# Example: 
num1 = int(input("Enter your first number:"))
num2 = int(input("Enter your second number:"))
print(num1 + num2)
```

2. To combine number with text:

```python
# Example: 
age = 25
#print("I am " + age + " years old.")

# Example: 
age = 25
print("I am " + str(age) + " years old.")
# note* You cannot directly concatenate a number with a string using +.
```

3. To Handle decimal value:

```python
# Example: use float()
price = "99.50"
price = float(price)

print(price + 10)
```

### Explicit and Implicit Type Conversion in Python:

Python supports two main types of conversion:

1. Explicit conversion — you tell Python to convert the data type.

Example 1: String to Integer:

```python
# Example: 
age = "25"
age = int(age)
print(age)
print(type(age))
# note* explicitly used int() to convert "25" from a string into an integer.
```

Example 2: Integer to Float:

```python
# Example: 
number = 10
result = float(number)
print(result) 
```

Example 3: Number to String:

```python
# Example: 
price = 500
text = str(price)
print("Price: " + text)
```

2. Implicit conversion — Python automatically converts the data type when appropriate.

```python
# Example: 
number = 10
decimal = 2.5
result = number + decimal
print(result)
print(type(result))
```

### How conversion affects calculations and comparisons:

This is especially important when working with user input because input() returns text (string) by default.


1. Conversion affects calculation:

```python
# Python performs calculations differently depending on the data type.
# Example: 
a = input("Enter first number: ")
b = input("Enter second number: ")
print(a + b)
# note* Because a and b are strings, so + joins the text instead of adding the numbers.

# Example 2: 
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
print(a + b)
# note* int() converts the strings into integers, allowing Python to perform numerical addition.
```

2. Conversion affects Mulitiplication:

```python
# Example: 
x = "5"
print(x * 3)
# note* "5" is a string, * 3 repeats the string.

# Example 2: 
x = int("5")
print(x * 3)
# note* conversion changes the meaning of the operation.
```

3. Conversion affects comparisions:

```python
# Example: 
age = input("Enter your age: ")
print(age > "18")
# note* If the input is "20", Python compares strings, not numbers.
# note* String comparisons are based on character ordering, which can produce results that are different from numerical comparisons.

# Example 2: 
print("20" > "100")
# note* This may look strange because numerically 20 is less than 100.
# note* The reason is that Python compares the characters in the strings rather than their numerical values.
```

4. Converting before comparison: 

```python
# Example: 
age = int(input("Enter your age: "))
print(age > 18)
# note* Python performs a numeric comparison.
```

5. Comparing number and string:

```python
# Example: 
x = 10  # Integer
y = "10"    # String
print(x == y)
# note* Although they appear to contain the same value, they have different types.

# Example: 
x = 10
y = int("10")
print(x == y)
```
