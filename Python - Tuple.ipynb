### Definition: 
A tuple in Python is a collection used to store multiple values in one variable. The important feature of a tuple is that it is immutable, meaning you cannot change, add, or remove its elements after creating it.

```python
# Example: 
fruits = ("apple", "banana", "mango")
print(fruits)

# Example 2:
numbers = (10, 20, 30)
print(numbers)
```

### Tuple Indexing: 
Tuple indexing is the process of accessing individual elements of a tuple using their index number. Python uses zero-based indexing, meaning the first element has index 0, the second has index 1, and so on.

```python
# Basic Example: 
fruits = ("Apple", "Banana", "Mango", "Orange")
print(fruits[0])
print(fruits[1])
print(fruits[2])
```

1. Positive and Neagtive indexing: 

```python
# Example: 
fruits = ("Apple", "Banana", "Mango", "Orange")
# here, 
#       fruits:             Apple   Banana  Mango   orange
#       positive index:       0       1       2       3
#       Neagative index:     -4      -3      -2      -1
print(fruits[-1])
print(fruits[-3])
print(fruits[0])
print(fruits[2])
```

2. Indexing with expression: 


```python
# You can also use a variable as an index.
# Example: 
numbers = (10, 20, 30, 40)
index = 2
print(numbers[index])
```

3. Index out of Range: 

```python
# If you try to access an index that doesn't exist, Python gives an IndexError.
# Example: 
numbers = (10, 20, 30)
# print(numbers[5])
```

### Tuple Length:
The length of a tuple means the total number of elements stored inside the tuple. Python provides the built-in len() function to find the length of a tuple.

1. Tuple Length with String: 

```python
# Example: 
fruits = ("Apple", "Banana", "Mango", "Orange")
print(len(fruits))
```

2. Tuple Lenght with numbers:

```python
# Example:
numbers = (10, 20, 30, 40, 50)
print(len(numbers))
```

3. Empty Tuple:

```python
# Example; 
data = ()
print(len(data))
```

4. Length of Tuple with different data types:

```python
# Example: 
student = ("James", 20, 85.5, True)
print(len(student))
```

5. Nested Tuple: 


```python
# A nested tuple is a tuple inside another tuple.
# Example: 
numbers = (10, 20, (30, 40, 50))
print(len(numbers))
```
