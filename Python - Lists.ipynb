## Python_Lists Notebook

### Definition: 
A list is a collection in Python that allows you to store multiple values in a single variable.

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(fruits)
```

### The Use of Collection
A collection allows you to group multiple related values together. This is useful when a program needs to work with many pieces of data.

```python
# Example: 
# note* variables use for each names to store.
student1 = "Peter"
student2 = "James"
student3 = "Henry"

# Example: 
# Instead: note* with list; one variables can store multiple data. 
students = ["Peter", "James", "Henry"]
print(students)
```

### Storing different data types.
Python lists can store different types of data in the same list. This is useful when related information has different data types.

```python
# Example: 
student = ["Yubraj", 20, 75.5, True]        # Here: "Yubaraj" - string (str)
print(type(student[0]))                     #        20        - integer (int)
print(type(student[1]))                     #        75.5      - float (float)
print(type(student[2]))                     #        true      - bool (bool)
print(type(student[3]))
```

### Accessing elements using positive and negative indexing.
In Python, indexing means accessing an individual element of a list using its position.

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(fruits [0])
print(fruits [3])

# Here: Positive index: 
#       "Apple"         - 0
#       "Banana"        - 1
#       "Mango"         - 2
#       "Orange"        - 3

# For neagtive index: 
print(fruits [-1])
print(fruits [-3])

# Here: Neagtive index: 
#       "Apple"         - -4
#       "Banana"        - -3
#       "Mango"         - -2
#       "Orange"        - -1
```

### Slicing Lists
List slicing means taking a portion of a list to create a new list containing selected elements.

```python
# Example: by positive index slicing:
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(fruits[1:4])
print(fruits [2:5])
```

1. Omitting the Start: 


```python
# This means start from the beginning:
# Example:
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(fruits[:3])
```

2. Omitting the stop:

```python
# This means start from the end:
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(fruits[2:])
```

3. Copying the entire list:

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(fruits[:])
```

4. Using negative slicing:

```python
# Negative indexes can also be used:
# Example; 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(fruits[-3:])
```

5. Reversing a list:

```python
# A step of (::-1) reverses:
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(fruits[::-1])
```

### Changing existing elements.
Python lists are mutable, which means you can change their elements after the list has been created.

1. Changing One Element:

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
fruits[2] = "Dates"
print(fruits)

# Here, the element at index 2 ("Mango") was replaced with "Dates".
```

2. Changing Multiple Elements:

```python
# Example: 
numbers = [10, 20, 30, 40, 50]
numbers[1:4] = [200, 300, 400]
print(numbers)

# Here, The elements at indexes 1, 2, and 3 were replaced by 200, 300, 400.
```

3. Changing an Element Using Negative Indexing:

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
fruits[-4] = "Avacado"
print(fruits)

# Here, the element at index -4 ("Banana") was replaced with "Avacado".
```

Changing the Data Type of an Element:

```python
# Example: 
data = ["Python", 20, 75.5]
data[1] = "Programming"
print(data)
```

### Adding and removing elements
Python lists are mutable, so you can add new elements and remove existing elements after creating a list.

1. Adding an Element with .append():

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
fruits.append("Avacado")
print(fruits)
```

2. Adding an Element at a Specific Position with .insert():

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
fruits.insert(1,"Cherry")
print(fruits)

# Here, "Cherry" is inserted at index 1 "Banana".
```

3. Adding Multiple Elements with .extend():

```python
# Example: 
fruits = ["Apple", "Banana"]
fruits.extend(["Cherry", "Dragonfruit"])
print(fruits)
```

4. Removing a Specific Value with .remove():

```python
# deletes the first matching value.
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
fruits.remove("Orange")
print(fruits)
```

5. Removing an Element by Index with .pop():

```python
# .pop() removes an element using its index and returns the removed value.
# Example:
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
remove = fruits.pop(2)
print(fruits)
print(remove)
```

6. Removing an Element with .del:

```python
# del can remove an element using its index.
# Example: 
fruits = ["Apple", "Banana", "Mango"]
del fruits[1]
print(fruits)
```

7. Removing All Elements with .clear():

```python
# .clear() removes everything from the list.
# Example: 
fruits = ["Apple", "Banana", "Mango"]
fruits.clear()
print(fruits)
```

### Finding the length of a list.
The length of a list means the number of elements contained in that list.

```python
# Exmaple: 
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(len(fruits))

# Example 2: 
numbers = [10, 20, 30, 40, 50]
print(len(numbers))
```

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(len(fruits))
print(fruits[len(fruits) - 1])
```

### Important list operations and methods.
Python provides many operations and methods for working with lists. They allow you to add, remove, search, sort, and manipulate elements.

1. Adding Elements: .append()

```python
# Example: 
fruits = ["Apple", "Banana"]
fruits.append("Mango")
print(fruits)
```

2. Removing Elements: .remove()

```python
# Example: 
fruits = ["Apple", "Banana", "Mango", "Orange"]
fruits.remove("Mango")
print(fruits)
```

3. Finding Infromation: .len()


```python
# Example: 
numbers = [10, 20, 30, 40]
print(len(numbers))
```

4. Sorting Lists: 

```python
# Ascending sort: 
# Example: 
numbers = [40, 10, 30, 20]
numbers.sort()
print(numbers)

# Descending sort:
# Exmaple: 
numbers = [40, 10, 30, 20]
numbers.sort(reverse=True)
print(numbers)
```

5. Reversing a List: .reverse()

```python
# Example; 
numbers = [1, 2, 3, 4]
numbers.reverse()
print(numbers)
```

6. Checking Membership: .in

```python
# Use in to check whether an element exists.
# Example: 
fruits = ["Apple", "Banana", "Mango"]
print("Apple" in fruits)
print("Orange" in fruits)
```

7. Combining Lists: (+)

```python
# Example: 
list1 = [1, 2, 3]
list2 = [4, 5, 6]
result = list1 + list2
print(result)
```

8. Repeating Lists: (*)

```python
# Exmaple: 
list1 = [1, 2, 3]
print(list1 * 3)
```

### Sorting, reversing, searching, counting, and clearing list data.

3. Searching a List: in()

4. Counting Elements: .count()

```python
# Exmaple: 
numbers = [10, 20, 10, 30, 10, 40]
print(numbers.count(10))

# Example 2: 
fruits = ["Apple", "Mango", "Apple", "Banana", "Apple"]
print(fruits.count("Apple"))
```

5. Clearing a List: .clear()

```python
# Example:
fruits = ["Apple", "Banana", "Mango"]
fruits.clear()
print(fruits)
```

### Experiments involving real-world collections such as students, marks, products, or tasks.

1. Students: 

```python
# Example: 
students = ["James", "Charles", "Hamilton", "Harry"]

print("Students:", students)
print("First student:", students[0])

students.append("Patrick")
print("After adding:", students)

students.remove("Hamilton")
print("After removing:", students)

print("Number of students:", len(students))
```

2. Students Marks: 

```python
# Example:
marks = [75, 88, 62, 95, 70]

print("Marks:", marks)
print("Highest mark:", max(marks))
print("Lowest mark:", min(marks))

marks.sort()
print("Ascending order:", marks)

marks.reverse()
print("Descending order:", marks)

print("Number of students:", len(marks))
```

3. Product Collection:

```python
# Example:
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]

print("Products:", products)

products.append("Printer")
print("After adding:", products)

products.remove("Mouse")
print("After removing:", products)

print("Is Laptop available?", "Laptop" in products)
print("Total products:", len(products))
```

4. Task List: 

```python
# Example: 
tasks = ["Study Python", "Complete homework", "Exercise"]

print("Tasks:", tasks)

tasks.append("Read a book")
print("New task added:", tasks)

tasks[0] = "Study Python lists"
print("Updated task:", tasks)

tasks.remove("Exercise")
print("After completing a task:", tasks)
```
