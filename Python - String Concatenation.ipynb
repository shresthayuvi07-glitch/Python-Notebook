## Python_String_Concatenation Notebook

### Difinition: 
Concatenation means joining two or more strings together to create a single string. 

```python
# Example: 
first = "Hello"
second = "Heaven"
result = first + second
print(result)
```

### Adding a Space:

```python
# Example: 
first = "Hello"
second = "Heaven"
result = first + " " + second 
print (result)
```

### Combining Text Values: 


Joining two or more string to create a single piece of text.

```python
# Example: 
first = "Hello"     # Using operator +
second = "Heaven"
result = first + " " + second 
print (result)

# Example 2:
city = "Vienna" 
country = "Austria" 
location = city + ", " + country
print(location)
```

### Combining Variables with Text:

1. Using Operator (+):

```python
# Example: 
name = "Mann" 
city = "London"
message = "my name is " + name + " and I live in " + city + "."
print (message)
```

2. Using F string:

```python
# Example: 
name = "Yubaraj"
city = "Kathmandu"
location = f"My name is {name}, and I live in {city}" 
print (location)
```

### Different approach for formatting text output: 


1. Using Commas with print():

```python
# Example; 
name = "Yubraj"
age = 25

print("Name:", name, "Age:", age)
# Note* Python automatically adds space between the values
```

2. Using string concatenation:

```python
# Example: 
name = "Yubraj"
age = 25

print("My name is " + name + " and I am " + str(age) + " years old.") 
# note* using +, all values are joined and must be string so, str() is necessary. 
```

3. Using Format: .format()

```python
# Example: 
name = "Mann"
age = 30 
print("my name is {} and I am {} years old.".format(name, age))
```
