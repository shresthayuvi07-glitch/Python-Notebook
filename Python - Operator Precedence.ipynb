## Python_Operator_Precedence Notebook

### Definition: 
Operator precedence is the set of rules that determines the order in which operators are evaluated in an expression.

```python
# Example: 
x= 5
y= 10
z= 15
print(z + (x*y))
```

### Precedence Order: 
1. Parentheses ()
2. Exponents ^ (or ** in some languages)
3. Multiplication *, Division /, Modulus %
4. Addition +, Subtraction -
5. Comparison operators (<, >, ==,)
6. Logical operators (&&, //, and, or,

### Why the order of operations matters
The order of operations matters because an expression can contain several operators, and the computer needs clear rules to decide which calculation to perform first.

```python
# Example: 
x= 5
y= 10
z= 20
print(z + (x*y))
```

### The general priority of parentheses, arithmetic operations, comparisons, and logical operations.
The general priority (precedence) of these operators determines which operation is performed first when an expression contains multiple types of operators.

1. Parentheses — highest priority

```python
# Parentheses are evaluated first.
# Example: 
x= 5
y= 10
z= 20
print(z + (x*y))
```

2. Arithmetic operations

```python
# Arithmetic operations generally happen before comparisons and logical operations.
# Example:
result = 10 + 5 * 2
print(result)
```

3. Comparison operations

```python
# Comparisons are normally performed after arithmetic operations.
# Example: 
result = 10 + 5 > 12
print(result)
```

4. Logical operations

```python
# Logical operators generally come after comparisons.
# Example: 
age = 20
has_id = True

result = age >= 18 and has_id
print(result)
```

### How parentheses can change the result of an expression.
Parentheses can change the result of an expression by forcing Python to perform the operations inside the parentheses before the operations outside them.

1. Addition and multiplication:

```python
# Without Parentheses: 
# Example:
result = 5 + 3 * 2
print(result)

# With Parentheses: 
# Example: 
result = (5 + 3) * 2
print(result)
```

2. Division and subtraction:

```python
# Without Parentheses:
# Example: 
result = 20 - 8 / 2
print(result)

# with Parentheses: 
# Example: 
result = (20 - 8) / 2
print(result)
```

3. Comparisons and logical operations

```python
# Parentheses can also make logical expressions clearer and can change how conditions are grouped.
# Example: 
x = 10
result = (x > 5) and (x < 20)
print(result)
```

### Why parentheses are useful:
1. Override normal operator precedence
2. Make the programmer's intention clear
3. Make complex expressions easier to read
4. Help prevent calculation mistakes
5. Can significantly change the final result
