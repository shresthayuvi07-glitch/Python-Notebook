### 1. Opening and Closing a File:

**Step 1.1** — Create a file to work with.

```python
f = open("sample.txt", "w")
f.write("Hello, this is line 1.\n")
f.close()
print("sample.txt created!")
```

**Step 1.2** — Add two more lines (still separate steps).

```python
f = open("sample.txt", "a")
f.write("This is line 2.\n")
f.close()
print("Added line 2.")
```

```python
f = open("sample.txt", "a")
f.write("And this is line 3.\n")
f.close()
print("Added line 3.")
```

**Step 1.3** — Open the file for reading.

```python
file = open("sample.txt", "r")
print("File opened. Is it closed?", file.closed)
```

**Step 1.4** — Read its content.

```python
content = file.read()
print(content)
```

**Step 1.5** — Close it, and check the **.closed** flag change.

```python
file.close()
print("Is it closed now?", file.closed)
```

If you forget **.close()**, the file may stay locked and changes might not be saved. Section 5 shows a safer way that closes it automatically.


### File Modes:

| Mode | Meaning |
|------|---------|
| **'r'** | Read (default). Errors if file doesn't exist. |
| **'w'** | Write. Creates if missing, **overwrites** if it exists. |
| **'a'** | Append. Creates if missing, adds to the end if it exists. |
| **'x'** | Create. Errors if the file already exists. |
| **'b'** | Binary mode add-on, e.g. **'rb'**, **'wb'**. |


**Step 2.1** — Write to **demo_modes.txt** for the first time with **'w'**.

```python
f = open("demo_modes.txt", "w")
f.write("First write.\n")
f.close()

with open("demo_modes.txt") as f:
    print(f.read())
```

**Step 2.2** — Use **'w'** again and watch it OVERWRITE the previous content.

```python
f = open("demo_modes.txt", "w")
f.write("Second write (this replaced the first!)\n")
f.close()

with open("demo_modes.txt") as f:
    print(f.read())
```

**Step 2.3** — Now use **'a'** and watch it APPEND instead.

```python
f = open("demo_modes.txt", "a")
f.write("This line was appended.\n")
f.close()

with open("demo_modes.txt") as f:
    print(f.read())
```

**Step 2.4** — Append one more line, one step at a time.

```python
f = open("demo_modes.txt", "a")
f.write("So was this one.\n")
f.close()

with open("demo_modes.txt") as f:
    print(f.read())
```

**Step 2.5** — Try **'x'** mode on a file that already exists — it should fail.

```python
try:
    f = open("demo_modes.txt", "x")
    f.write("This will not run.")
    f.close()
except FileExistsError as e:
    print("Got the expected error:", e)
```

### Reading Files — Three Different Methods

**Step 3.1** — **.read()**: the whole file as one string.

```python
with open("sample.txt", "r") as f:
    whole_text = f.read()
print(repr(whole_text))
```

**Step 3.2** — **.readline()**: read just the first line.

```python
with open("sample.txt", "r") as f:
    line1 = f.readline()
print(repr(line1))
```

**Step 3.3** — Call **.readline()** again on a fresh open — see the second line.

```python
with open("sample.txt", "r") as f:
    line1 = f.readline()
    line2 = f.readline()
print("Line 1:", repr(line1))
print("Line 2:", repr(line2))
```

**Step 3.4** — **.readlines()**: get every line as a list.

```python
with open("sample.txt", "r") as f:
    lines = f.readlines()
print(lines)
```

### File Pointers: **seek()** and **tell()**

**Step 6.1** — Open the file and check the starting cursor position.

```python
f = open("sample.txt", "r")
print("Start position:", f.tell())
```

**Step 6.2** — Read 5 characters and see the cursor move.

```python
chunk = f.read(5)
print("Read:", repr(chunk))
print("Position after reading 5 chars:", f.tell())
```

**Step 6.3** — Use **seek(0)** to rewind to the start.

```python
f.seek(0)
print("Position after seek(0):", f.tell())
```

**Step 6.4** — Read again from the start — same 5 characters.

```python
chunk_again = f.read(5)
print("Reading again from start:", repr(chunk_again))
f.close()
```

### Working with Paths

**Step 7.1** — **os.path**: get the absolute path.

```python
import os
path = "sample.txt"
print("Absolute path:", os.path.abspath(path))
```

**Step 7.2** — Check existence and type.

```python
print("File exists?", os.path.exists(path))
print("Is a file?", os.path.isfile(path))
print("Is a directory?", os.path.isdir(path))
```

**Step 7.3** — Break the path into pieces.

```python
print("File name only:", os.path.basename(path))
print("Directory only:", os.path.dirname(os.path.abspath(path)))
print("Extension:", os.path.splitext(path))
```

**Step 7.4** — Get the file size.

```python
print("File size (bytes):", os.path.getsize(path))
```

### Creating, Checking, and Deleting Files & Folders

**Step 8.1** — Create a folder.

```python
import os
os.makedirs("demo_folder", exist_ok=True)
print("Folder created:", os.path.isdir("demo_folder"))
```

**Step 8.2** — Create a file inside that folder.

```python
with open(os.path.join("demo_folder", "note.txt"), "w") as f:
    f.write("A note inside demo_folder.\n")

print("Contents of demo_folder:", os.listdir("demo_folder"))
```

**Step 8.3** — Rename the file.

```python
os.rename(os.path.join("demo_folder", "note.txt"),
          os.path.join("demo_folder", "renamed_note.txt"))
print("After rename:", os.listdir("demo_folder"))
```

**Step 8.4** — Delete the file.

```python
os.remove(os.path.join("demo_folder", "renamed_note.txt"))
print("After delete:", os.listdir("demo_folder"))
```

### Exception Handling with Files

**Step 9.1** — Try to open a file that doesn't exist.

```python
try:
    f = open("does_not_exist.txt", "r")
    data = f.read()
    f.close()
except FileNotFoundError:
    print("That file doesn't exist — handled gracefully instead of crashing.")
```

### Summary

| Task | Tool |
|------|------|
| Open/close a file | **open()**, **.close()**, or **with open(...) as f:** |
| Read | **.read()**, **.readline()**, **.readlines()**, or loop over the file |
| Write | **.write()**, **.writelines()** |
| Modes | **'r'**, **'w'**, **'a'**, **'x'**, **'+'**, **'b'** |
| Move the cursor | **.seek()**, **.tell()** |
| Paths | **os.path** or **pathlib.Path** |
| Manage filesystem | **os.makedirs**, **os.listdir**, **os.rename**, **os.remove**, **os.rmdir** |
| Handle errors | **try** / **except FileNotFoundError** / **finally** |
| CSV | **csv.reader**, **csv.writer**, **csv.DictReader**, **csv.DictWriter** |
| JSON | **json.load**, **json.dump**, **json.loads**, **json.dumps** |
| Binary | **'rb'** / **'wb'** modes |

**Golden rule:** prefer **with open(...) as f:** over manual **open()**/**close()**.
