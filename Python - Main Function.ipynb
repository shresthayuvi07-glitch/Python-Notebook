### Purpose of Main Function Structure:

The function Structure is called, what information it receives, what it does, and what result it may return.


```python
# Example:
def add_numbers(a, b): 
    result = a + b 
    return result
answer = add_numbers(10, 30)
print(answer)
# Here: 
# Def               - Tells python to create a function
# add_numbers       - Give the Funtion name
# (Parameters)      - Recive input value
# :                 - Indicates the begin of function body
# print             - give an output
```

### The role of Function Definition and Function Execution:

### -Function Definition: 
It is where create and describe what a function do. Function Execution: It happens when you call the Function.

```python
# Example:
def add_number(a, b):   # Function Definition
    return a + b
result = add_number(20, 40)  # Function Execution
print(result)
```

-Organizing a Programs with Main Function improve readability and structure, Why? Because with out Main funciton it become difficult to understand and maintain and it can also separate task into different function. Plus it makes debugging easier.

### -Function Execution:

1. Function is defined but not called,

```python
# Example: 
def text():
    print("Hello Yubaraj")
print("How is your python coding Going?")
```

2. Function is defined and called,

```python
# Example:  
def text():
    print("Hello Yubaraj")
text()
print("Nothin is going good in Python")   
```

3. Function with Parameter,

```python
# Example:  
def add(a, b):
    result = a + b
    print(result)
print("Adding the Numbers")
add(10, 50)
print("Calculation Done.")
```
