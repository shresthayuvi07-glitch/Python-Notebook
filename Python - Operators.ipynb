### Arithmetic operators and mathematical calculations:

### Definition:
Arithmetic operators are symbols used to perform mathematical calculations in Python. They work with numeric values such as integers (int) and decimal numbers (float).

| Operator | Name           |   Example | Result |
| -------- | -------------- | --------: | -----: |
| `+`      | Addition       |  `10 + 5` |   `15` |
| `-`      | Subtraction    |  `10 - 5` |    `5` |
| `*`      | Multiplication |  `10 * 5` |   `50` |
| `/`      | Division       |  `10 / 5` |  `2.0` |
| `//`     | Floor Division | `10 // 3` |    `3` |
| `%`      | Modulus        |  `10 % 3` |    `1` |
| `**`     | Exponentiation |  `2 ** 3` |    `8` |


2. Addition

```python
#The (+) operator adds two or more numbers.
# Example:
a = 10
b = 5
result = a + b
print(result)
```

2. Subtraction: 

```python
#The (-) operator subtracts one number from another.
# Example:
a = 20
b = 8
print(a - b)
```

3. Multiplication:

```python
# The (*) operator multiplies values.
# Example: 
price = 50
quantity = 4
total = price * quantity
print(total)
```

4. Division:

```python
#The (/) operator performs division. Python normally returns a float, even when the result is a whole number.
# Example: 
result = 10 / 2
print(result)
```

5. Floor Division: 

```python
#The (//) operator performs division and removes the decimal portion by rounding toward negative infinity.
# Example: 
result = 10 // 3
print(result)
```

6. Modulus:

```python
# The (%) operator gives the remainder after division.
# Example:
result = 10 % 3
print(result)

# note* This is especially useful for checking whether a number is even or odd:
# Example: 
number = 7
if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

7. Exponentiation

```python
# The (**) operator raises a number to a power.
# Example: 
result = 2 ** 4
print(result)
```

8. Combining Arithmetic Operators

```python
# Python allows multiple operations in one expression.
# Example: 
result = 10 + 5 * 2
print(result)
```

9. Mathematical Calculation Using Variables

```python
# Can store values in variables and use them in calculations.
# Example: 
length = 10
width = 5

area = length * width

print("Area =", area)
```

### Comparison Operators:

Comparison operators are used to compare two values. The result of a comparison is always a Boolean value:
True or False: and they are commonly used in if statements, loops, and decision-making programs.

| Operator | Meaning                  | Example    | Result |
| -------- | ------------------------ | ---------- | ------ |
| `==`     | Equal to                 | `10 == 10` | `True` |
| `!=`     | Not equal to             | `10 != 5`  | `True` |
| `>`      | Greater than             | `10 > 5`   | `True` |
| `<`      | Less than                | `5 < 10`   | `True` |
| `>=`     | Greater than or equal to | `10 >= 10` | `True` |
| `<=`     | Less than or equal to    | `5 <= 10`  | `True` |


1. Equal to (==)

```python
# The == operator checks whether two values are equal.
# Example: 
x = 10
y = 10
print(x == y)
# note* if the values are different; the result will be False. 
# Remember* = is assigns a value, while == is compares values.
```

2. Not Equal (!=)

```python
# The != operator checks whether two values are different.
# Example: 
age = 20
print(age != 18)
```

3. Greater Than (>)

```python
# Checks whether the value on the left is greater than the value on the right.
# Example: 
print(15 > 10)
```

4. Less Than (<)

```python
# Checks whether the left value is smaller than the right value.
# Example: 
print(5 < 10)
```

5. Greater Than or Equal To (>=)

```python
# Checks whether a value is greater than or equal to another value.
# Example: 
score = 50
print(score >= 50)
```

6. Less Than or Equal To (<=)

```python
# Checks whether a value is less than or equal to another value.
# Example: 
age = 18
print(age <= 18)
```

### Logical Operators

Logical operators are used to combine or modify conditions. They work with Boolean values (True and False) and return a Boolean result.

| Operator | Meaning                                    | Example            |
| -------- | ------------------------------------------ | ------------------ |
| `and`    | True if **both** conditions are true       | `x > 5 and x < 20` |
| `or`     | True if **at least one** condition is true | `x > 10 or x < 5`  |
| `not`    | Reverses the Boolean result                | `not(x > 5)`       |


1. (and) Operator:

```python
# The and operator returns True only when both conditions are true.
# Example: 
age = 25
print(age >= 18 and age <= 60)
```

Truth Table:
| Condition 1 | Condition 2 | `and` |
| ----------- | ----------- | ----- |
| True        | True        | True  |
| True        | False       | False |
| False       | True        | False |
| False       | False       | False |


```python
# Example: 
username = "admin"
password = "1234"
if username == "admin" and password == "1234":
    print("Login successful")
# note* both condition must be satisfied.
```

2. (or) Operator:

```python
# The or operator returns True when at least one condition is true.
# Example: 
age = 16
print(age < 18 or age > 60)
```

Truth Table: 
| Condition 1 | Condition 2 | `or`  |
| ----------- | ----------- | ----- |
| True        | True        | True  |
| True        | False       | True  |
| False       | True        | True  |
| False       | False       | False |


```python
# Example: 
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend")
    
# note* Only one of the two conditions needs to be true.
```

3. (not) Operator:

```python
# The not operator reverses a Boolean result.
# Example: 
x = 10
print(not(x > 5))
```

### Identity Operators

Identity operators are used to check whether two variables refer to the same object in memory. Python has two identity operators: 
| Operator | Meaning                                                      | Example      |
| -------- | ------------------------------------------------------------ | ------------ |
| `is`     | Checks whether two variables refer to the same object        | `x is y`     |
| `is not` | Checks whether two variables do not refer to the same object | `x is not y` |


1. (Is) Operator:

```python
# The is operator checks whether two variables refer to the same object.
# Example: 
x = [1, 2, 3]
y = x
print(x is y)
```

2. (is not) Operator

```python
# The is not operator checks whether two variables refer to different objects.
# Example: 
x = [1, 2, 3]
y = [1, 2, 3]
print(x is not y)
```

### Membership Operators

Membership operators are used to check whether a value exists inside a collection, such as a string, list, tuple, set, or dictionary. Python has two membership operators:
| Operator | Meaning                               | Example              |
| -------- | ------------------------------------- | -------------------- |
| `in`     | Checks whether a value exists         | `"a" in "apple"`     |
| `not in` | Checks whether a value does not exist | `"z" not in "apple"` |


1. (in) Operator

```python
# The in operator checks whether a value exists in a sequence or collection.
# Example: 
fruits = ["apple", "banana", "orange"]
print("banana" in fruits)
print("mango" in fruits)
```

2. (not in) Operator

```python
# The not in operator checks whether a value does not exist in a collection.
# Example: 
fruits = ["apple", "banana", "orange"]
print("mango" not in fruits) 
print ("Dragonfruits" not in fruits)
```

### Membership Operators with Strings

```python
#You can check whether a character or piece of text exists inside a string.
# Example: 
text = "Python Programming"
print("Python" in text)
print("Java" in text)
```

### Membership Operators with Lists

```python
# Example:
numbers = [10, 20, 30, 40, 50]
print(30 in numbers)
print(100 in numbers)
```

### Membership Operators with Tuples

```python
# Example: 
colors = ("red", "green", "blue")
print("green" in colors)
print("yellow" not in colors)
```

### Membership Operators with Sets

```python
# Example:
numbers = {10, 20, 30, 40}
print(20 in numbers)
print(50 not in numbers)
```

### Membership Operators with Dictionaries

```python
# With dictionaries, (in) checks whether a value is present as a key.
# Example:
student = {
    "name": "Yubraj",
    "age": 20,
    "course": "Python"
}
print("name" in student)
print("address" in student)
```

### Using Membership Operators in if Statements

```python
# Membership operators are often used for decision-making.
# Example: 
fruits = ["apple", "banana", "orange"]
fruit = "banana"
if fruit in fruits:
    print("Fruit is available")
else:
    print("Fruit is not available")
```

### Practical Example: Checking a Username

```python
# Example: 
allowed_users = ["admin", "john", "alice"]
username = "john"
if username in allowed_users:
    print("Access granted")
else:
    print("Access denied")
```
