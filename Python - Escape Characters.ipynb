## Python_Escape_Characters Notebook: 

### Definition: 
It is used inside strings to represent things that are difficult or impossible to type directly. In Python, escape characters usually begin with a backslash (\).

### It's Uses: 

1. Create a new line
2. Insert a tab space
3. Include quotation marks inside a string
4. Insert a backslash
5. Create special formatting within text

### Creating New Lines and Tabs: 


 Creating New lines: \n

```python
# Example: 
print("Hello\nWorld") 
# note* \n tells Python to move the following text to a new line.
```

 Tabs: \t

```python
# Example: 
print ("Name: \t Peter")
print ("Age: \t 24")
# note* \t inserts a tab space.
```

### Displaying Quotation marks inside text: 


```python
# Quotation Marks: \"
# Exmaple: 
print("He said \"Python is easy.\"")
# note* The \" tells Python that the quotation mark is part of the text rather than the end of the string.
```

### Combining multiple escape characters: 

```python
# Example: 
print("Name:\tYubaraj\nAge:\t25\nCountry:\tNepal")
# note* You can use multiple escape characters in single string.
```

### Common Mistake involving special character:


1. Forgetting to escape quotation marks: 

```python
# Example: 
#print("He said "Hello"") # note* Python thinks the string ends before "Hello".

# Example: 
print("He said \"Hello\"")
```

2. Mixing quotation marks incorrectly:

```python
# Example: 
#name = "Yubaraj' 
# note* The string starts with " but ends with '.

# Example: 
name = "Yubaraj"
name = 'Peter'
```

3. Forgetting the backslash in escape character:

```python
# Example: 
print ("Hello n World")

# Example: 
print ("Hello \nWorld")
# note* The \ is necessary for Python to recognize \n as an escape sequence
```

4. Forgetting that \t creates a tab:

```python
# Example: 
print ("Peter\tParker")
```
