## Python_Dictionary Notebook

### Definition: 
A dictionary in Python is a collection used to store data in key-value pairs. Each piece of data has a key and a corresponding value.

```python
# Example: 
student = {
    "name": "James",
    "age": 20,
    "marks": 85
}
print(student)

# note* Set and dictionary both use curly bracket to store data. However, dictionary store data using a corresponding value as key- value pairs.
```

### Accessing Dictionary Value:
Values are accessed using their keys.

```python
# Example: 
student = {
    "name": "James",
    "age": 20,
    "Location": "Vienna, Austria"
}
print(student["name"])
print(student["Location"])
```

### Adding Items:
A new key-value pair can be added to a dictionary.

```python
# Example: 
student = {
    "name": "James",
    "age": 20,
    "Location": "Vienna, Austria"
}
student["Height"]= 5.11
print(student)
```

### Changing Values:
Dictionaries are mutable, which means their values can be changed.

```python
# Example:
student = {
    "name": "James",
    "age": 20,
    "Location": "Vienna, Austria"
}
student["name"]= "Patrick"
student["age"]= 25
print(student)
```

### Removing Items:
The del keyword can remove an item.

```python
# Example:
student = {
    "name": "James",
    "age": 20,
    "Location": "Vienna, Austria"
}
del student["age"]
print(student)
```

### Finding Dictionary Lenght: 
The len() function returns the number of key-value pairs.

```python
# Example: 
student = {
    "name": "James",
    "age": 20,
    "Location": "Vienna, Austria"
}
student["height"]= 5.11
print(student)
print(len(student))
```

### Looping Through a Dictionary:
You can use a for loop to access dictionary items.

```python
# Example: 
student = {
    "name": "James",
    "age": 20,
    "Location": "Vienna, Austria"
}
for key, value in student.items():
    print(key, ":", value)
```

### Dictionary with Different Data Types:
A dictionary can contain different types of values.

```python
# Example:
student = {
    "name": "James",
    "age": 20,
    "Location": "Vienna, Austria",
    "height": 5.11,
    "student": True
}
print(student)
```

### Adding Dictionaries inside Dictionary: 

```python
# Example: 
students = {
    "student1":{
    "name": "James",
    "age": 24,
    "location": "Vienna, Austria"
}, 
    "student2":{
    "name": "Patrick", 
    "age": 25,
    "location": "Paris, France"
}, 
    "student3":{
    "name": "Peter", 
    "age": 30, 
    "location": "Helsinki, Finland"
}}
print(students)
```

### Dictionary Method: 

| Method         | Purpose                           |
| -------------- | --------------------------------- |
| `keys()`       | Returns all keys                  |
| `values()`     | Returns all values                |
| `items()`      | Returns key-value pairs           |
| `get()`        | Gets a value using a key          |
| `update()`     | Adds or updates items             |
| `pop()`        | Removes a specified item          |
| `popitem()`    | Removes the last inserted item    |
| `clear()`      | Removes all items                 |
| `copy()`       | Creates a copy                    |
| `setdefault()` | Gets a value or creates a default |
| `fromkeys()`   | Creates a dictionary from keys    |


Dictionary Method with Example: 


```python
# Example; 
student = {
    "name": "Patrick", 
    "age": 25,
    "location": "Paris, France"
}
print(student.keys())
print(student.values())
print(student.items())
print(student.get("name"))
student.update({"Height": 5.11})
print(student)
```
