### Definition: 
An if statement is a decision-making statement in Python. It is used to execute a block of code only when a specified condition is True.

```python
# Example: 
age = 20
if age >= 18:
    print("You are an adult.")

# if condition: 
#   The colon : after the condition is required, and the code inside the if block must be indented.
```

1. Using if with Comparison Operators:

```python
# Example: 
marks = 35
if marks >= 40:
    print("You passed.")
else:
    print("You failed.")
```

2. Using if with Multiple Conditions:

```python
# You can use logical operators such as and, or, and not.
# Example:
age = 25
has_id = True
if age >= 18 and has_id:
    print("Entry allowed.")
else:
    print("Entry Declined.") 
```

### Variables in Condition Statement: 

1. Using a Variable with ==:


```python
# The == operator checks whether two values are equal.
# Example: 
marks = 80
if marks == 80:
    print("You scored 80 marks.")
else:
    print("You didnot.")
```

2. Using a String Variable:

```python
# Variables containing text can also be used in conditions.
# Example:
name = "Jude"
if name == "Jude":
    print("Welcome Jude!")
else: 
    print("Sorry! Wrong person.")
```

3. Using Multiple Variables:

```python
# You can use more than one variable in a condition.
# Example: 
age = 24
country = "Nepal"

if age >= 18 and country == "Nepal":
    print("Condition is satisfied.")
else:
    print("Condition is not satisfied.")
```

4. Using User Input in a Condition:

```python
# Variables are often used to store information entered by the user.
# Example: 
age = int(input("Enter your age: "))
if age >= 18:
    print("You can vote.")
else:
    print("You are underage.")
```

### The Elif Keyword in Python:
The elif keyword in Python means "else if". It is used when you want to check multiple conditions one after another. Furthermore, Python checks the conditions from top to bottom. As soon as it finds a condition that is True, it executes that block and skips the remaining conditions.

```python
# Example: 
marks = 90

if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```

### The Else Keyword in Python: 
The else keyword is used in Python to execute a block of code when the condition in an if statement is False.

```python
# Example: 
age = 15
if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

```python
# Example 2:
age = int(input("Enter your age: "))
if age >= 18:
    print("You can vote.")
else:
    print("You cannot vote yet.")
```

### Python Logical Operator: 
Logical operators in Python are used to combine multiple conditions and produce a result of either True or False.

1. and Operator:

```python
# The and operator returns True only when both conditions are True.
# Example:
age = 15
has_id = True
if age >= 18 and has_id:
    print("Entry allowed.")
else:
    print("Entry Declined.")
```

2. or Operator:

```python
# The or operator returns True when at least one condition is True.
# Example: 
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It is the weekend.")
else:
    print("Sorry!")
```

3. not Operator:

```python
# The not operator reverses the Boolean result.
# Exmaple:
is_raining = False
if not is_raining:
    print("You can go outside.")
```

### Combining Logical Operator: 

```python
# Example: 
age = 25
has_id = True
is_member = False
if age >= 18 and (has_id or is_member):
    print("Access granted.")
else:
    print("Access Denied.")
```
