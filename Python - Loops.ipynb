## Python_Loops Notebook

### Definition:
In Python, loops are used to repeat a block of code multiple times. They help you avoid writing the same code over and over.

### Types of Loops in Python:

#### For Loop in Python:
A for loop in Python is used to repeat a block of code for each item in a sequence or collection. It is commonly used with lists, tuples, strings, dictionaries, and the range() function.

```python
# Example:
for i in range(5):
    print(i)
    
# For loop syntax: 
# for variable in sequence:
```

1. Using range():


```python
# Example: 
for i in range(5):
    print(i)
```

2. Looping through a List: 

```python
# Example:
fruits = ["Apple", "Banana", "Mango"]
for fruit in fruits:
    print(fruit)
```

3. Loping through String:

```python
# Example: 
word = "Python"
for letter in word:
    print(letter)
```

4. Using range(start, stop, step)

```python
# Example: 
for i in range(1, 10, 2):
    print(i)
# Here,   
# 1 = starting value
# 10 = stopping value, not included
# 2 = step/increment
```

5. for Loop with if statement:

```python
# A for loop can be combined with conditions:
# Example:
for i in range(1, 6):
    if i % 2 == 0:
        print(i)
```

#### While Loop in Python: 
A while loop in Python is used to repeat a block of code as long as a specified condition is True. It is useful when you do not know exactly how many times the loop should run.

```python
# Example: 
count = 1
while count <= 5:
    print(count)
    count += 1

# While loop syntax:
# while condition:
```

1. Printing a Message reapeatly:

```python
# Example: 
i = 1
while i <= 3:
    print("Hello Python")
    i += 1
```

2. While Loop with User Input:

```python
# Example: 
number = int(input("Enter a number: "))
while number != 5:
    print("You entered:", number)
    number = int(input("Enter another number: "))
# note* The loop continues until the user enters 5.
```

### Break Statement in while Loop: 
break is used to stop the loop immediately.

```python
# Example: 
count = 1
while count <= 10:
    if count == 5:
        break
    print(count)
    count += 1
```

### Continue Statement in a While Loop: 
continue skips the current iteration and moves to the next iteration.

```python
# Example: 
count = 0
while count < 10:
    count += 1
    if count == 3:
        continue
    if count == 5:
        continue

    print(count)
```
