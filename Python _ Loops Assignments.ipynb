### For Loop:

1. Print numbers from 1 to 10:


```python
# Coding: 
i = 1
for i in range(1, 11):
    print(i)
# note* use range(start, stop, step)
```

2. Print even number from 1 to 20:

```python
# Coding: 
i = 0 
for i in range(1, 21):
    if i % 2 == 0:
        print (i)
```

3. Print odd number from 1 to 20:

```python
# Coding: 
i = 0 
for i in range(1, 20):
    if i % 2 != 0:
        print (i)
```

4. Print all element in the list:

```python
# Coding: 
fruits = ["Apple", "Banana", "Mango", "Cherry", "Apricot"]
for fruit in fruits:
    print(fruit)
```

### While Loop:

1. Print numbers from 1 to 10:


```python
# Coding: 
count = 1
while count <= 10:
    print (count)
    count += 1
```

2. Print numbers from 10 to 1: 


```python
# Coding: 
count = 10
while count >= 1:
    print (count)
    count -= 1
```

3. Print Even numbers upto 50: 


```python
# Coding: 
count = 2
while count <= 50 :
    print (count)
    count+= 2
```

4. print sum of first 10 numbers: 

```python
# Coding: 
num = 1
total = 0
while num <= 10:
    total = total + num
    num = num + 1
print("Sum =", total)
```
