## Python_User_Input Notebook:

### Defination: 
It is information that user provides to a program while the program is running. The input() function is commonly used to receive input. 

```python
# Example: 
name = input("Enter Your Name:")
print ("Hello", name)
```

### User Input need in Interactive Programs:


1. User Input makes programs flexible:


1. With String:

```python
# Example:
name = "Mann"           # without user input: fixed 
print("Hello:", name)

name = input("Enter Your Name:") # with user input : changeable
print ("Hello:", name)
```

2. With Integer: 

```python
# Example: 
a = 10 
b = 5       # Without user input
print (a+b) 

a = int(input("Enter Your First Number:"))
b = int(input("Enter Your Second Number:"))     # with user input
print(a+b)
```

 2. User Input allows Personalization: 

```python
# Example: 
name = input("Enter Your Name:")
city = input("Enter Your City Name:")
print("Name:", name)
print("City:", city)
```

3. User Input and Decision Making: 


```python
# Example: 
age = int(input("Enter Your Age:"))
if age >= 20: 
    print("You are an Adult.")
else: 
    print("You are Minor.")
```

### Accepting Infromation from User: 


1. Text or String:

```python
# Example:
name = input("Enter Your Name:")
city = input("Enter Your City Name:")
print("Name:", name)
print ("city:", city)
```

2. Integer or Number:

```python
# Example: 
age = int(input("Enter Your Age:"))
grade = int(input("Enter Your Grade:"))
print ("Age:", age)
print("Grade:", grade)
```

3. Decimal Number:

```python
# Example:
price = float(input("Enter the Price:"))
print ("Price:", price)
```

4. Multiple Inputs:

```python
# Example:
name = input("Enter Your Name:")
age = int(input("Enter Your Age:"))
city = input("Enter Your City Name:")
print("Name:", name)
print ("Age:", age)
print ("city:", city)
```

5. Accepting two numbers for Calculations: 

```python
# Example: 
num1 = int(input("Enter Your First Number:"))
num2 = int(input("Enter Your Second Number:"))
total = num1 + num2
print("Total:", total)
```

### Default type returned by User Input: 

By input() function, the value returned is always as a string (str) by default. For integer, use int() and for float, use float().

```python
# Example: 
age = int(input("Enter Your Age:"))
height = float(input("Enter Your Height:"))
marks = int(input("Enter Your Marks:"))
print(type(age))
print(type(height))
print(type(marks))
```

### Convert User Input into Numeric Type:

### Handling common User input related mistake:

1. Forgetting that input() function always return a string.

```python
# Example: 
#age = input("Enter your Age:")   # Python give error.
#print(age + 5)

age = int(input("Enter Your Age:"))
print(age)      # Correct way
```

2. Entering invalid Numeric Data:

```python
# Example: 
age = int(input("Enter Your Age:"))
print(age)
# Note* if 'twenty' is entered, Python give error.

try: 
    age = int(input("Enter Your Age:"))
    print("Your age is:", age)
except ValueError:
    print("Please Enter a Valid Number.")
```

3. Using float() for decimal number:

```python
# Example: 
price = float(input("Enter Your Price:"))
print (price)
# note* donot use int() for decimal number. 
```

4. Empty Input:

```python
# Example: 
name = input("Enter Your Name:")
if name == "":
    print("Name cannot be empty.")
else:
    print("Hello", name)
```

5. Input outside an Expected Range: 

```python
# Example: 
age = int(input("Enter Your Age:"))
if 5 <= age <= 120:
    print("Valid Age")
else:
    print("Please Enter an age between 5 to 120")   
```

6. Handling Division by Zero:

```python
# Example: 
num1 = int(input("Enter Your First Number:"))
num2 = int(input("Enter Your Second Number:"))
if num2 == 0:
    print("Cannot be divided by Zero.")
else:
    print("Result:", num1/num2) 
```
