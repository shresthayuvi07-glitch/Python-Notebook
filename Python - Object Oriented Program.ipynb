### **The __init__() Method:**

Every class can define an __init__() method.

It is executed automatically when a new object is created.

Its main purpose is to give an object its initial property values and perform setup needed at object creation time.

### **Basic pattern**

```python
class ClassName:
    def __init__(self, value):
        self.value = value
```

When we call ClassName(...), Python runs __init__() for the new object.

````python
# Example:
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Lilly", 25)

print(p1.name, p1.age)
print(p1.name)
````

### **Why use __init__()?:**

Without __init__(), properties often have to be assigned manually after each object is created.

Using __init__() makes object creation cleaner because the values are supplied at the moment the object is created.

````python
# Exmaple:
# Without __init__(): values are assigned manually
class PersonWithoutInit:
    pass

p1 = PersonWithoutInit()
p1.name = "Tank"
p1.age = 25

print(p1.name)
print(p1.age)
````

````python
# Example:
# With __init__(): values are supplied immediately
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Licht", 20)

print(p1.name)
print(p1.age)
````

### **The self Parameter:**
self refers to the **current object**.

It is used inside instance methods to access properties and other methods that belong to that particular object. For example:

```python
self.name
```
means: the name property of the object currently using this method.

````python
# Example:
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("Jerry", 25)
p1.greet()
````

### **Why is self important?:**
Suppose two objects come from the same class:

- one object has the name "Lilly",
- another has the name "Steve".

The same method can be called by both objects. self tells Python which object's data the method should work with.

````python
# Example:
class Person:
    def __init__(self, name):
        self.name = name

    def printname(self):
        print(self.name)

p1 = Person("Lilly")
p2 = Person("Steve")

p1.printname()
p2.printname()
````

### **Methods in a Class:**
Methods are functions defined inside a class. They describe what objects created from that class can do.

### **Deleting a method from a class:**

The PDF also demonstrates removing a method from a class with **del**.

After the method is deleted from the class, objects of that class can no longer call it.

````python
# Example:
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello!")

p1 = Person("Emil")
p1.greet()

del Person.greet

try:
    p1.greet()
except AttributeError as error:
    print("After deletion:", error)
````

### **Inheritance:**

**Inheritance** lets one class reuse behavior from another class.

The existing class is usually called the **parent/base class**, while the new class is called the **child/derived class**.

The inheritance diagram in the PDF lists:

- Single inheritance
- Multilevel inheritance
- Multiple inheritance
- Hierarchical inheritance
- Hybrid inheritance

### **Single Inheritance:**

**One parent → one child**

The child class inherits behavior from one parent class.

````python
# Example: 
class Parent:
    def show(self):
        print("Parent class")

class Child(Parent):
    def display(self):
        print("Child class")

c = Child()
c.show()
c.display()
````

### **Multilevel Inheritance:**

**Parent → child → grandchild**

A class inherits from a class that already inherited from another class.

The last class in the chain can access inherited behavior from the earlier levels.

````python
# Example:
class Grandparent:
    def house(self):
        print("Grandparent house")

class Parent(Grandparent):
    def car(self):
        print("Parent car")

class Child(Parent):
    def bike(self):
        print("Child bike")

c = Child()
c.house()
c.car()
c.bike()
````

### **Multiple Inheritance:**

**Multiple parents → one child**

A child class inherits from more than one parent class.

````python
# Example:
class Father:
    def skill(self):
        print("Father skill")

class Mother:
    def skill(self):
        print("Mother skill")

class Child(Father, Mother):
    pass

c = Child()
c.skill()
````

For a class with several inheritance paths, Python needs a predictable search order.

You can inspect it with:

```python

```

````python
print(Child.mro())
````

### **Hierarchical Inheritance:**

**One parent → multiple children**

Two or more child classes inherit from the same parent class.

````python
# Example: 
class Animal:
    def eat(self):
        print("Animal can eat")


class Dog(Animal):
    def bark(self):
        print("Dog barks")


class Cat(Animal):
    def meow(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.eat()
dog.bark()

cat.eat()
cat.meow()
````

### **Polymorphism:**

**Polymorphism** means **"having more than one form."**

It allows the same method or operation to behave differently in different situations.

The PDF first gives the general OOP classification:

1. **Static polymorphism** — compile-time polymorphism
2. **Dynamic polymorphism** — run-time polymorphism

### **Static Polymorphism — general OOP theory:**

It associates compile-time polymorphism with:

- method overloading,
- operator overloading,
- selecting the matching operation at compile time.

The PDF states that Python does not support traditional compile-time method overloading in the same way because Python is dynamically typed and method resolution is not based on strict compile-time signatures.

A later method definition with the same name replaces an earlier one rather than creating a traditional overload set.

````python
# In Java:
# Exmaple:
class Calculator {

    int add(int a, int b) {
        return a + b;
    }

    int add(int a, int b, int c) {
        return a + b + c;
    }
}
````

````python
# Example:
Calculator c = new Calculator();

System.out.println(c.add(10, 20));
System.out.println(c.add(10, 20, 30));
````

***Python does not support traditional method overloading based on method signatures like Java or C++.***

````python
# Example: 
class Test:
    def add(self, a, b):
        return a + b

    # This definition replaces the previous method with the same name.
    def add(self, a, b, c):
        return a + b + c

t = Test()
print(t.add(1, 2))
````

````python
# Example: 
class Calculator:
    
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):
        return a + b + c


c = Calculator()

print(c.add(10, 20))
````
