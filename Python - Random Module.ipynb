## Python_Random_Module Notebook

### Definition: 
The Python random module is a built-in module that provides functions for generating pseudo-random values.

### The purpose of Random Module
It is useful when a program needs to make random choices or generate unpredictable looking values, such as:

1. Rolling dice 
2. Choosing a random student
3. Generating random numbers
4. Creating simple games
5. Shuffling a list
6. Selecting random items

### Importing Random Module
The import statement is used to bring a module into your program.

```python
# Exmaple: 
import random
```

### Using a Module
The random module can generate a random number:

```python
# Example:
import random
number = random.randint(1, 10)
print(number)
```

### Generating random decimal
Python's built-in random module can generate random numbers. You first import it:

1. Generating a Random Decimal Value:

```python
# Use random.random() to generate a random floating-point (decimal) number.
# Example: 
import random
number = random.random()
print(number)
```

2. Generating a Random Decimal Within a Range:

```python
# Use random.uniform(a, b) when you want a random decimal between two values.
# Example: 
import random
number = random.uniform(10, 20)
print(number)
```

### Generating Integer Value

1. Generating a Random Integer:

```python
# Example: 
import random
number = random.randint(1, 10)
print(number)
```

2. Generating an Integer Using .randrange():

```python
# random.randrange() generates a random value from a range.
# Example: 
import random
number = random.randrange(1, 10)
print(number)
```

### Selecting random items from collections.
Python's random module provides functions for selecting random elements from collections such as lists, tuples, and strings.

1. Using random.choice()

```python
# The choice() function selects one random item from a sequence.
# Example: 
import random
students = ["James", "Charles", "Harry", "Kiara"]
selected = random.choice(students)
print("Selected student:", selected)
```

2. Selecting a Random Item from a Tuple

```python
# Example: 
import random
colors = ("Red", "Blue", "Green", "Yellow")
print(random.choice(colors))
```

3. Selecting a Random Character from a String

```python
# Example: 
import random
word = "PYTHON"
print(random.choice(word))
```

### Selecting multiple items.
Use random.sample() when you want to select multiple unique items.

```python
# Example: 
import random
students = ["Knight", "Sahara", "Hendrix", "Patrick", "Peter"]
selected = random.sample(students, 2)
print(selected)
```

### Shuffling Data in Python
Shuffling means randomly changing the order of items in a collection. Python's random module provides the shuffle() function for this purpose.

```python
# Example: 
numbers = [1, 2, 3, 4, 5]
print("Original numbers:", numbers)
random.shuffle(numbers)
print("Shuffle numbers", numbers)

# Example: 
students = ["Knight", "Sahara", "Hendrix", "Patrick", "Peter"]
print("Original Students List:", students)
random.shuffle(students)
print("Shuffle Students List:", students)
```
